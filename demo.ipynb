**edit-reel** is a free, local-first open-source AI video editor and auto subtitles generator for vertical short-form video (Instagram Reels, TikTok, and YouTube Shorts).

This interactive notebook demonstrates:
1. **Speech-Protected Trimming**: Whisper word-boundary extraction & silence removal
2. **Acoustic Forced Alignment**: Word-by-word timestamp synchronization
3. **Tactile Paper & Objects Rendering**: Compositing talking heads on warm paper `#E2DDD1` with hanging price tags, 3D coin stacks, and academic notebook cutoffs
4. **Broadcast Audio Mastering**: -11.5 LUFS loudness normalization with ambient music bed & motion SFX
